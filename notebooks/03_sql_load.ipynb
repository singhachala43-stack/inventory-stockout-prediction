```python
import pandas as pd
from sqlalchemy import create_engine

# Load our cleaned dataset from Day 2
df = pd.read_csv('../data/inventory_data_cleaned.csv', parse_dates=['Date'])
print(f"Loaded {len(df)} rows from CSV")

# Connect to MySQL
# Format: mysql+pymysql://username:password@host/database_name
engine = create_engine("mysql+pymysql://root:NewPassword123!@localhost/inventory_project")

# Push the DataFrame into MySQL as a table
df.to_sql('inventory_sales', con=engine, if_exists='replace', index=False)

print("Data loaded into MySQL table 'inventory_sales'")
```
